```python
import numpy as np
from scipy.sparse import csr_matrix

d = np.array([3, 4, 5, 7, 2, 6])     # data
r = np.array([0, 0, 1, 1, 3, 3])     # rows
c = np.array([2, 4, 2, 3, 1, 2])     # cols

csr = csr_matrix((d, (r, c)), shape=(4, 5))
print(csr.toarray())
```

```python
import numpy as np
from scipy.sparse import csc_matrix

d = np.array([3, 4, 5, 7, 2, 6])     
r = np.array([0, 0, 1, 1, 3, 3])     
c = np.array([2, 4, 2, 3, 1, 2])     

csc = csc_matrix((d, (r, c)), shape=(4, 5)) 
print(csc.toarray())
```

```python
import numpy as np
from scipy.sparse import coo_matrix

d = np.array([3, 4, 5, 7, 2, 6]) 
r = np.array([0, 0, 1, 1, 3, 3]) 
c = np.array([2, 4, 2, 3, 1, 2]) 

coo = coo_matrix((d, (r, c)), shape=(4, 5))
print(coo.toarray())
```

- dok_matrix (Dictionary of Keys)

```python
import numpy as np
from scipy.sparse import dok_matrix

dok = dok_matrix((4, 5))
dok[0, 2] = 3
dok[0, 4] = 4
dok[1, 2] = 5
dok[1, 3] = 7
dok[3, 1] = 2
dok[3, 2] = 6

print(dok.toarray())
```

- dia_matrix (Diagonal Matrix)

```python
import numpy as np
from scipy.sparse import dia_matrix

data = np.array([[3, 5, 6, 7]])  
offsets = np.array([0])         

dia = dia_matrix((data, offsets), shape=(4, 5))
print(dia.toarray())
```
